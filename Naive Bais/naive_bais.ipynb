```python
# Import libraries
import numpy as np

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Load dataset
data = load_diabetes()

X = data.data
y = data.target

# Convert continuous target into 5 categories
y_bins = np.digitize(
    y,
    bins=np.quantile(y, [0.2, 0.4, 0.6, 0.8])
)

# Train test split
X_train, X_test, y_train, y_test, y_train_bins = train_test_split(
    X, y, y_bins,
    test_size=0.2,
    random_state=42,
    stratify=y_bins
)

# Feature scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Create model
model = GaussianNB()

# Train model
model.fit(X_train, y_train_bins)

# Predict categories
y_pred_bins = model.predict(X_test)

# Convert predicted categories into numerical values
class_means = {
    category: y_train[y_train_bins == category].mean()
    for category in np.unique(y_train_bins)
}

y_pred = np.array([class_means[c] for c in y_pred_bins])

# Evaluation
print("MSE:", mean_squared_error(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
